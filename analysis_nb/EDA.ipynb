```python
plt.hist(df["Amount"], bins=50, edgecolor="black", log=True)
plt.xlabel("Amount")
plt.ylabel("Frequency")
plt.title("Amount histogram(log)")
plt.show()
```

```python
ecdf = df["Amount"].value_counts(normalize=True).sort_index().cumsum()
ecdf.plot(marker='.', linestyle='none')
plt.xlabel('Amount')
plt.ylabel('ECDF')
plt.title('ECDF plot of Amount')
plt.show()
```

```python
plt.hist(df[df["Amount"]<50]["Amount"], bins=50, edgecolor="black")
plt.xlabel("Amount")
plt.ylabel("Frequency")
plt.title("Amount histogram(log)")
plt.show()
```

```python
sns.boxplot(x=df[df["Amount"]<50]["Amount"])
plt.show()
```

```python
ecdf = df[df["Amount"]<50]["Amount"].value_counts(normalize=True).sort_index().cumsum()
ecdf.plot(marker='.', linestyle='none')
plt.xlabel('Amount')
plt.ylabel('ECDF')
plt.title('ECDF plot of Amount')
plt.show()
```

```python
plt.hist(df[(df["Class"]==1)]["Amount"], bins=50, edgecolor="black", log=True)
plt.xlabel("Amount")
plt.ylabel("Frequency")
plt.title("Amount histogram(log)")
plt.show()
plt.hist(df[df["Amount"]<2500]["Amount"], bins=50, edgecolor="black", log=True)
plt.xlabel("Amount")
plt.ylabel("Frequency")
plt.title("Amount histogram(log)")
plt.show()
```
