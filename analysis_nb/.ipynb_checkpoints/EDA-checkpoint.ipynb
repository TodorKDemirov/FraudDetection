```python
df["Amount"].quantile([0.25, 0.5, 0.75, 0.9, 0.95])
```

```python
plt.hist(df["Amount"], bins=50, edgecolor="black", log=True)
plt.xlabel("Amount")
plt.ylabel("Frequency")
plt.title("Amount histogram(log)")
plt.show()
```

```python
sns.histplot(df["Amount"], bins=50, edgecolor="black", kde=True)
plt.xlabel("Amount")
plt.ylabel("Frequency")
plt.title("Amount histogram(KDE)")
plt.show()
```

```python
ecdf = df["Amount"].value_counts(normalize=True).sort_index().cumsum()
ecdf.plot(marker='.', linestyle='none')
plt.xlabel('Amount')
plt.ylabel('ECDF')
plt.title('ECDF plot of Amount')
plt.show()
```

```python
sns.boxplot(x=df["Amount"])
plt.show()
```

```python
plt.hist(df[df["Amount"]<50]["Amount"], bins=50, edgecolor="black")
plt.xlabel("Amount")
plt.ylabel("Frequency")
plt.title("Amount histogram(log)")
plt.show()
```

```python
sns.boxplot(x=df[df["Amount"]<50]["Amount"])
plt.show()
```

```python
ecdf = df[df["Amount"]<50]["Amount"].value_counts(normalize=True).sort_index().cumsum()
ecdf.plot(marker='.', linestyle='none')
plt.xlabel('Amount')
plt.ylabel('ECDF')
plt.title('ECDF plot of Amount')
plt.show()
```

```python
df[df["Class"]==1]["Amount"]
```

```python
plt.hist(df[(df["Class"]==1) & (df["Amount"]>100)]["Amount"], bins=50, edgecolor="black")
plt.xlabel("Amount")
plt.ylabel("Frequency")
plt.title("Amount histogram(log)")
plt.show()
```

```python
df[df["Class"]==1]["Amount"].count()
```

```python
df[(df["Class"]==1) & (df["Amount"]>0)]["Amount"].count()
```

```python
df.loc[df["Amount"] == 0, "Amount_categorical"] = "zero_amount"
df.loc[(df["Amount"] > 0) & (df["Amount"] < 20), "Amount_categorical"] = "small"
df.loc[(df["Amount"] >= 20) & (df["Amount"] < 200), "Amount_categorical"] = "medium"
df.loc[df["Amount"] >= 200, "Amount_categorical"] = "large"
```
